```python
# Define a mapping for matching vacancy dataset job_title_unified
title_mapping = {
    'Business Intelligence Analyst': 'Business Analyst',
    'Data Scientist': 'Data Scientist',
    'Software Engineer': 'Software Engineer',
    'Cloud Engineer': 'Cloud Engineer',
    'Machine Learning Engineer': 'Machine Learning Engineer',
    'AI Researcher': 'Data Scientist'
}
```

```python
# Apply the mapping to the 'recommended_job_1' column and filter out specific job titles
raw['recommended_job_1'] = raw['recommended_job_1'].replace(title_mapping)
raw = raw[~raw['recommended_job_1'].isin([
    'Cybersecurity Analyst',
    'Frontend Developer',
    'Backend Developer',
    'DevOps Engineer',
    'Full Stack Developer',
    'Security Engineer'
])]
```

```python
# Display the count of each unique value in the 'recommended_job_1' column after mapping and filtering
raw['recommended_job_1'].value_counts()
```

After mapping and filtering 2 specializations are missing: Data Analyst and Data Engineer. And class disbalance is present

```python
# Over sampling to 300 samples per class and ensuring all classes are represented
all_classes = [
    'Software Engineer',
    'Business Analyst', 
    'Machine Learning Engineer',
    'Cloud Engineer',
    'Data Scientist',
    'Data Analyst',   
    'Data Engineer', 
]

df = balance_dataset(raw, target_count=300, all_classes=all_classes)
```
